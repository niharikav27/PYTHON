# Regression plot
-> primarily intended to add a visual guide that helps to emphasize patterns in a dataset during exploratory data analyses   
1. regplot() -> axis level function
2. lmplot() -> figure level function

```python
import seaborn as sns
```

```python
df = sns.load_dataset("penguins")
```

```python
df.head()
```

### 1. regplot()

```python
sns.regplot(data=df,x='flipper_length_mm',y='body_mass_g')
```

```python
# hue parameter is not present in regplot.

sns.regplot(data=df,x='flipper_length_mm',y='body_mass_g',hue='sex') #Error
```

2. lmplot()

```python
sns.lmplot(data=df,x='flipper_length_mm',y='body_mass_g',hue="sex")
```

-> residual plot -> calculating error in regression model.

```python
#residplot()
sns.residplot(data=df,x='flipper_length_mm',y='body_mass_g')
```

# Multigraph

1. pairplot / pairgrid
2. jointplot / jointgrid

```python
df.head()
```

1. pairplot()

```python
sns.pairplot(df)
```

```python
# hue
sns.pairplot(data=df, hue="species")
```

-> palette parameter : Change color schemes     
 = {"viridis", "Set2", "plasma", "coolwarm"}

```python
# palette
sns.pairplot(data=df, hue="species",palette="Set2")
```

-> diag_kind: Choose between "kde" or "hist" for diagonal.

```python
sns.pairplot(data=df, hue="species", diag_kind="hist", palette="viridis")
```

```python
sns.pairplot(data=df, hue="species", markers=["o", "s", "D"], palette="coolwarm")
```

-> var : select the columns on which pairplot should be made.

```python
# var
sns.pairplot(data=df, vars=["bill_length_mm", "bill_depth_mm", "flipper_length_mm"], hue="species")
```

```python
# corner=True
sns.pairplot(data=df, hue="species", palette="Set2",corner=True)
```

```python
# pairgrid()
a = sns.PairGrid(df,hue='species')
a.map(sns.scatterplot)
```

-> In PairGrid you have the customization that you can change any plot type

```python
# map_offdiag ans map_diag

a = sns.PairGrid(df,hue='species')
a.map_offdiag(sns.scatterplot)
a.map_diag(sns.histplot)
```

```python
a = sns.PairGrid(df,hue='species')
a.map_offdiag(sns.histplot)
a.map_diag(sns.violinplot)
```

```python
# map_upper and map_lower
a = sns.PairGrid(df,hue='species')
a.map_diag(sns.histplot)
a.map_upper(sns.scatterplot)
a.map_lower(sns.kdeplot)
```

```python
# vars
a = sns.PairGrid(df,hue='species',vars=['flipper_length_mm','body_mass_g'])
a.map_diag(sns.histplot)
a.map_upper(sns.scatterplot)
a.map_lower(sns.kdeplot)
```

2. jointplot

```python
# kind = scatter
sns.jointplot(data=df,x='flipper_length_mm',y='body_mass_g')
```

```python
# kind = kde
sns.jointplot(data=df, x="bill_length_mm", y="bill_depth_mm", hue="species", kind="kde")
```

```python
# kind =hist
sns.jointplot(data=df,x='flipper_length_mm',y='body_mass_g',kind='hist')
```

```python
# hue
sns.jointplot(data=df,x='flipper_length_mm',y='body_mass_g',kind='hist',hue='sex')
```

JointGrid()

```python
a = sns.JointGrid(df,x='flipper_length_mm',y='body_mass_g')
a.plot(sns.scatterplot,sns.boxplot)
```

```python
a = sns.JointGrid(df,x='flipper_length_mm',y='body_mass_g')
a.plot(sns.kdeplot,sns.histplot)
```
